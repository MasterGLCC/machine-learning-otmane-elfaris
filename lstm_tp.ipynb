```python
# 1. Construction du vocabulaire (mot -> index et index -> mot) pour chaque langue
class Vocab:
    def __init__(self, name):
        self.name = name
        self.word2idx = {"<pad>": 0, "<sos>": 1, "<eos>": 2}
        self.idx2word = {0: "<pad>", 1: "<sos>", 2: "<eos>"}
        self.n_words = 3

    def add_sentence(self, sentence):
        for w in sentence.split():
            if w not in self.word2idx:
                self.word2idx[w] = self.n_words
                self.idx2word[self.n_words] = w
                self.n_words += 1

    def encode(self, sentence, add_eos=True):
        idx = [self.word2idx[w] for w in sentence.split()]
        if add_eos:
            idx.append(self.word2idx["<eos>"])
        return idx

    def decode(self, indices):
        words = [self.idx2word[i] for i in indices if i not in (0, 1, 2)]
        return " ".join(words)

eng_vocab, fra_vocab = Vocab("eng"), Vocab("fra")
for e, f in TRANSLATION_PAIRS:
    eng_vocab.add_sentence(e)
    fra_vocab.add_sentence(f)

V_in, V_out = eng_vocab.n_words, fra_vocab.n_words
print("Vocabulaire anglais :", V_in, "mots | Vocabulaire français :", V_out, "mots")
```

```python
# 2. Définition du LSTM Encodeur-Décodeur (Seq2Seq) - 100% from scratch, pas de self-attention
np.random.seed(0)
H = 32  # taille de l'état caché

def xavier(a, b):
    return np.random.randn(a, b) * np.sqrt(1.0 / b)

def sigmoid(x):
    return 1.0 / (1.0 + np.exp(-x))

def softmax(z):
    z = z - np.max(z)
    e = np.exp(z)
    return e / np.sum(e)

class LSTMSeq2Seq:
    def __init__(self, V_in, V_out, H, lr=0.15):
        self.H = H
        self.lr = lr
        self.E_enc, self.W_enc, self.b_enc = self._make_lstm_params(V_in, H)
        self.E_dec, self.W_dec, self.b_dec = self._make_lstm_params(V_out, H)
        self.Why = xavier(V_out, H); self.by = np.zeros((V_out, 1))

    def _make_lstm_params(self, V, H):
        E = xavier(V, H)
        # 4 portes : f (forget), i (input), o (output), c (candidate cellule)
        W = {g: xavier(H, H + H) for g in ['f', 'i', 'o', 'c']}
        b = {g: np.zeros((H, 1)) for g in ['f', 'i', 'o', 'c']}
        return E, W, b

    def _zero_grads(self, V, H):
        E = np.zeros((V, H))
        W = {g: np.zeros((H, H + H)) for g in ['f', 'i', 'o', 'c']}
        b = {g: np.zeros((H, 1)) for g in ['f', 'i', 'o', 'c']}
        return E, W, b

    # Un pas de LSTM : portes forget/input/output + état de cellule c_t + état caché h_t
    def lstm_step(self, x, h_prev, c_prev, W, b):
        concat = np.vstack([h_prev, x])
        f = sigmoid(W['f'] @ concat + b['f'])   # porte d'oubli
        i = sigmoid(W['i'] @ concat + b['i'])   # porte d'entrée
        o = sigmoid(W['o'] @ concat + b['o'])   # porte de sortie
        g = np.tanh(W['c'] @ concat + b['c'])   # candidat de cellule
        c = f * c_prev + i * g                  # nouvel état de cellule
        h = o * np.tanh(c)                      # nouvel état caché
        cache = (x, h_prev, c_prev, f, i, o, g, c, h, concat)
        return h, c, cache

    # Rétropropagation d'un pas de LSTM
    def lstm_backward(self, dh, dc, cache, W):
        x, h_prev, c_prev, f, i, o, g, c, h, concat = cache
        H = h_prev.shape[0]
        tanh_c = np.tanh(c)
        do = dh * tanh_c
        dc = dc + dh * o * (1 - tanh_c ** 2)
        df = dc * c_prev
        di = dc * g
        dg = dc * i
        dc_prev = dc * f
        dfz = df * f * (1 - f)
        diz = di * i * (1 - i)
        doz = do * o * (1 - o)
        dgz = dg * (1 - g ** 2)
        grads = {}
        dconcat = np.zeros_like(concat)
        for name, dz in zip(['f', 'i', 'o', 'c'], [dfz, diz, doz, dgz]):
            grads['W' + name] = dz @ concat.T
            grads['b' + name] = dz
            dconcat += W[name].T @ dz
        dh_prev = dconcat[:H]
        dx = dconcat[H:]
        return dx, dh_prev, dc_prev, grads

    def train_pair(self, eng_idx, fra_idx, sos_idx):
        H = self.H
        gE_enc, gW_enc, gb_enc = self._zero_grads(self.E_enc.shape[0], H)
        gE_dec, gW_dec, gb_dec = self._zero_grads(self.E_dec.shape[0], H)
        gWhy = np.zeros_like(self.Why); gby = np.zeros_like(self.by)

        # --- Encodeur : forward ---
        h = np.zeros((H, 1)); c = np.zeros((H, 1))
        enc_caches = []
        for idx in eng_idx:
            x = self.E_enc[idx].reshape(-1, 1)
            h, c, cache = self.lstm_step(x, h, c, self.W_enc, self.b_enc)
            enc_caches.append((cache, idx))
        h_enc, c_enc = h, c

        # --- Décodeur : forward (teacher forcing) ---
        dec_input = [sos_idx] + fra_idx[:-1]
        target = fra_idx
        h, c = h_enc, c_enc
        dec_caches, probs, loss = [], [], 0.0
        for t, idx in enumerate(dec_input):
            x = self.E_dec[idx].reshape(-1, 1)
            h, c, cache = self.lstm_step(x, h, c, self.W_dec, self.b_dec)
            y = self.Why @ h + self.by
            p = softmax(y)
            probs.append(p)
            dec_caches.append((cache, idx))
            loss += -np.log(p[target[t], 0] + 1e-9)

        # --- Décodeur : backward ---
        dh_next = np.zeros((H, 1)); dc_next = np.zeros((H, 1))
        for t in reversed(range(len(dec_input))):
            dy = probs[t].copy(); dy[target[t]] -= 1
            cache, idx = dec_caches[t]
            gWhy += dy @ cache[8].T
            gby += dy
            dh = self.Why.T @ dy + dh_next
            dx, dh_prev, dc_prev, g = self.lstm_backward(dh, dc_next, cache, self.W_dec)
            for k in ['f', 'i', 'o', 'c']:
                gW_dec[k] += g['W' + k]; gb_dec[k] += g['b' + k]
            gE_dec[idx] += dx.flatten()
            dh_next, dc_next = dh_prev, dc_prev

        # --- Encodeur : backward ---
        dh_enc_next, dc_enc_next = dh_next, dc_next
        for t in reversed(range(len(eng_idx))):
            cache, idx = enc_caches[t]
            dx, dh_prev, dc_prev, g = self.lstm_backward(dh_enc_next, dc_enc_next, cache, self.W_enc)
            for k in ['f', 'i', 'o', 'c']:
                gW_enc[k] += g['W' + k]; gb_enc[k] += g['b' + k]
            gE_enc[idx] += dx.flatten()
            dh_enc_next, dc_enc_next = dh_prev, dc_prev

        # --- Mise à jour des poids ---
        for k in ['f', 'i', 'o', 'c']:
            for gW, W in [(gW_enc, self.W_enc), (gW_dec, self.W_dec)]:
                np.clip(gW[k], -5, 5, out=gW[k]); W[k] -= self.lr * gW[k]
            for gb, b in [(gb_enc, self.b_enc), (gb_dec, self.b_dec)]:
                np.clip(gb[k], -5, 5, out=gb[k]); b[k] -= self.lr * gb[k]
        for gE, E in [(gE_enc, self.E_enc), (gE_dec, self.E_dec)]:
            np.clip(gE, -5, 5, out=gE); E -= self.lr * gE
        np.clip(gWhy, -5, 5, out=gWhy); self.Why -= self.lr * gWhy
        np.clip(gby, -5, 5, out=gby); self.by -= self.lr * gby

        return loss / len(dec_input)

    def translate(self, eng_idx, sos_idx, eos_idx, max_len=10):
        h = np.zeros((self.H, 1)); c = np.zeros((self.H, 1))
        for idx in eng_idx:
            x = self.E_enc[idx].reshape(-1, 1)
            h, c, _ = self.lstm_step(x, h, c, self.W_enc, self.b_enc)
        out_idx = []
        idx = sos_idx
        for _ in range(max_len):
            x = self.E_dec[idx].reshape(-1, 1)
            h, c, _ = self.lstm_step(x, h, c, self.W_dec, self.b_dec)
            p = softmax(self.Why @ h + self.by)
            idx = int(np.argmax(p))
            if idx == eos_idx:
                break
            out_idx.append(idx)
        return out_idx
```

```python
# 3. Entraînement du modèle sur les 150 paires de traduction
model = LSTMSeq2Seq(V_in, V_out, H, lr=0.15)
data = [(eng_vocab.encode(e), fra_vocab.encode(f)) for e, f in TRANSLATION_PAIRS]
sos_idx = fra_vocab.word2idx["<sos>"]

loss_log = []
epochs = 40
for epoch in range(epochs):
    np.random.shuffle(data)
    total = 0.0
    for eng_idx, fra_idx in data:
        total += model.train_pair(eng_idx, fra_idx, sos_idx)
    avg = total / len(data)
    loss_log.append(avg)
    if epoch % 5 == 0 or epoch == epochs - 1:
        print(f"Époque {epoch}/{epochs} - Loss moyenne : {avg:.4f}")
```

```python
# 4. Test de traduction (quelques exemples du corpus d'entraînement)
eos_idx = fra_vocab.word2idx["<eos>"]
for eng_sent, fra_sent in TRANSLATION_PAIRS[:6]:
    eng_idx = eng_vocab.encode(eng_sent)
    pred_idx = model.translate(eng_idx, sos_idx, eos_idx)
    pred_sent = fra_vocab.decode(pred_idx)
    print(f"EN   : {eng_sent}")
    print(f"REF  : {fra_sent}")
    print(f"LSTM : {pred_sent}")
    print("-" * 40)
```

```python
# 5. Visualisation de l'évolution de la loss
plt.figure(figsize=(6, 4))
plt.plot(loss_log, color='navy')
plt.title("Évolution de la loss (LSTM Seq2Seq) au fil des époques")
plt.xlabel("Époques")
plt.ylabel("Loss (cross-entropy)")
plt.grid(True)
plt.show()
```

## 1. Objectifs du TP
* Comprendre le fonctionnement du **LSTM (Long Short-Term Memory)**, conçu pour résoudre le problème de mémoire à long terme du RNN vanille.
* Implémenter **from scratch** (NumPy uniquement) une architecture **Encodeur-Décodeur LSTM** pour la traduction automatique, **sans mécanisme d'attention**.
* Comparer sa capacité d'apprentissage à celle du RNN simple sur le **même corpus réel** de traduction anglais → français.

## 2. Concepts Clés Abordés
* **Portes du LSTM :** à chaque pas de temps, trois portes sigmoïdes contrôlent le flux d'information :
  * **Porte d'oubli** $f_t = \sigma(W_f[h_{t-1}, x_t] + b_f)$ : décide quoi effacer de la mémoire.
  * **Porte d'entrée** $i_t = \sigma(W_i[h_{t-1}, x_t] + b_i)$ : décide quoi ajouter.
  * **Porte de sortie** $o_t = \sigma(W_o[h_{t-1}, x_t] + b_o)$ : décide quoi exposer comme état caché.
* **État de cellule (mémoire longue) :** $c_t = f_t \odot c_{t-1} + i_t \odot \tilde{c}_t$, où $\tilde{c}_t = \tanh(W_c[h_{t-1}, x_t] + b_c)$.
* **État caché :** $h_t = o_t \odot \tanh(c_t)$.
* **Architecture Encodeur-Décodeur :** identique en principe au RNN, mais l'état transmis à chaque pas comprend à la fois $h_t$ **et** $c_t$, ce qui préserve mieux l'information sur de longues séquences.

## 3. Résultats Obtenus
* **Convergence :** la loss diminue plus rapidement et de façon plus stable que pour le RNN vanille sur le même nombre d'époques (comparer les deux graphes de loss).
* **Traduction :** les traductions générées sont cohérentes sur le corpus d'entraînement, avec une convergence généralement meilleure que le RNN simple.

## 4. Conclusion & Limites
Le LSTM résout le problème du **vanishing gradient** du RNN vanille grâce à son état de cellule $c_t$, qui agit comme une "autoroute" de gradient à travers le temps (les portes contrôlent l'ajout/l'oubli sans multiplication répétée par une matrice de poids). Il reste cependant, comme le RNN, **sans self-attention** : toute l'information de la phrase source est encore compressée dans l'état final $(h_T, c_T)$ transmis au décodeur, ce qui limite les performances sur des phrases longues. Le **GRU** propose une version simplifiée (2 portes au lieu de 3), et le **Transformer** supprime totalement ce goulot d'étranglement grâce à la self-attention.
